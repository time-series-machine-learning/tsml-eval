# The Canonical Interval Forest (CIF) Classifier for Time Series Classification

Matthew Middlehurst, James Large and Anthony Bagnall. IEEE International Conference on
Big Data, 2020.
[Paper](https://doi.org/10.1109/BigData50022.2020.9378424)

The paper introduces CIF, an interval based classifier using the catch22 features, and
HIVE-COTE with CIF replacing TSF (HC-CIF), and evaluates them on 112 UCR and 26 UEA
datasets.

The results for the paper are in the
[canonical-interval-forest-2020](https://github.com/time-series-machine-learning/canonical-interval-forest-2020)
repository:

- [Results](https://github.com/time-series-machine-learning/canonical-interval-forest-2020/tree/main/results)
  for the univariate and multivariate comparisons, with links to the detailed results
  and the code used
- [Running CIF](https://github.com/time-series-machine-learning/canonical-interval-forest-2020#running-cif)
  with `aeon` or the `tsml` code used in the paper
- [Example notebook](https://github.com/time-series-machine-learning/canonical-interval-forest-2020/blob/main/cif.ipynb)
  using CIF

CIF and catch22 are available in the [aeon](https://github.com/aeon-toolkit/aeon)
Python package.
