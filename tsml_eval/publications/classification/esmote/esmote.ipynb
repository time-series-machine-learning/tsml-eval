# e-SMOTE: A Train Set Rebalancing Algorithm for Time Series Classification

Chuanhang Qiu, Matthew Middlehurst, Christopher Holder and Anthony Bagnall.
International Workshop on Advanced Analytics and Learning on Temporal Data (AALTD),
2025.
[Paper](https://doi.org/10.1007/978-3-032-15535-1_1)

The paper introduces e-SMOTE, an elastic distance based oversampling method for
imbalanced time series classification, evaluated on an archive of 76 binary univariate
datasets with a class ratio of at least 9:1 in the train set.

The datasets and code are in the
[AALTD2025imbalance](https://github.com/LinGinQiu/AALTD2025imbalance)
repository:

- [Datasets](https://github.com/LinGinQiu/AALTD2025imbalance/tree/main/UCR_Imbalanced_9_1)
  in `.ts` format, also available from
  [Zenodo](https://doi.org/10.5281/zenodo.18641021)
- [How the archive was constructed](https://github.com/LinGinQiu/AALTD2025imbalance#how-the-benchmark-was-constructed)
- [e-SMOTE implementation](https://github.com/LinGinQiu/AALTD2025imbalance/blob/main/esmote.py)
  using `aeon`
