# HIVE-COTE 2.0: a new meta ensemble for time series classification

Matthew Middlehurst, James Large, Michael Flynn, Jason Lines, Aaron Bostrom and Anthony
Bagnall. Machine Learning, 2021.
[Paper](https://doi.org/10.1007/s10994-021-06057-9)

The paper introduces HIVE-COTE 2.0 (HC2), a meta ensemble of the Temporal Dictionary
Ensemble, the Diverse Representation Canonical Interval Forest, the Arsenal and the
Shapelet Transform Classifier, and compares it to the state of the art on 112 UCR and 26
UEA datasets.

The results for the paper are in the
[hivecote2-2021](https://github.com/time-series-machine-learning/hivecote2-2021)
repository:

- [Results](https://github.com/time-series-machine-learning/hivecote2-2021/tree/main/results)
  for the univariate and multivariate comparisons, the ablative study and the
  ensembling comparison, with links to the full prediction files
- [Running HC2](https://github.com/time-series-machine-learning/hivecote2-2021#running-hc2)
  with `aeon` or the `tsml` version used in the paper
- [Example notebook](https://github.com/time-series-machine-learning/hivecote2-2021/blob/main/hivecote2.ipynb)
  using HC2

HC2 and its components are available in the
[aeon](https://github.com/aeon-toolkit/aeon) Python package.
