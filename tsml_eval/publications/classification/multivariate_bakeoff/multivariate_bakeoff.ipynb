# The great multivariate time series classification bake off

Alejandro Pasos Ruiz, Michael Flynn, James Large, Matthew Middlehurst and Anthony
Bagnall. Data Mining and Knowledge Discovery, 2021.
[Paper](https://doi.org/10.1007/s10618-020-00727-3)

The paper reviews multivariate time series classification algorithms and compares them
on 26 equal length datasets from the UEA archive.

The results for the paper are in the
[multivariate-bakeoff-2021](https://github.com/time-series-machine-learning/multivariate-bakeoff-2021)
repository:

- [Results](https://github.com/time-series-machine-learning/multivariate-bakeoff-2021/tree/main/results)
  for 12 classifiers on the 26 UEA datasets
- [Running the classifiers](https://github.com/time-series-machine-learning/multivariate-bakeoff-2021#running-the-classifiers)
  in `tsml` and `aeon`

Most of the classifiers are available in the
[aeon](https://github.com/aeon-toolkit/aeon) Python package.
