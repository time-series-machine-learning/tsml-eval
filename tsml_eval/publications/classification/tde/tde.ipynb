# The Temporal Dictionary Ensemble (TDE) Classifier for Time Series Classification

Matthew Middlehurst, James Large, Gavin Cawley and Anthony Bagnall. European
Conference on Machine Learning and Principles and Practice of Knowledge Discovery in
Databases (ECML PKDD), 2020.
[Paper](https://doi.org/10.1007/978-3-030-67658-2_38)

The paper introduces TDE, a dictionary based classifier that combines advances from
earlier dictionary methods with Gaussian process based parameter selection, and
HIVE-COTE with TDE replacing BOSS (HC-TDE).

The results for the paper are in the
[temporal-dictionary-ensemble-2020](https://github.com/time-series-machine-learning/temporal-dictionary-ensemble-2020)
repository:

- [Results](https://github.com/time-series-machine-learning/temporal-dictionary-ensemble-2020/tree/main/results)
  for the dictionary based classifiers and the HIVE-COTE variants, with links to the
  detailed results and the code used
- [Running TDE](https://github.com/time-series-machine-learning/temporal-dictionary-ensemble-2020#running-tde)
  with `aeon` or the `tsml` code used in the paper
- [Example notebook](https://github.com/time-series-machine-learning/temporal-dictionary-ensemble-2020/blob/main/tde.ipynb)
  using TDE

TDE is available in the [aeon](https://github.com/aeon-toolkit/aeon) Python package.
