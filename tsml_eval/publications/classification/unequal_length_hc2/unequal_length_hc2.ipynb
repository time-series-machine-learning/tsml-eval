# Classifying unequal length time series with HIVE-COTE 2.0

Matthew Middlehurst and Anthony Bagnall. 2026 11th International Conference on Big Data
Analytics (ICBDA), 2026.

The paper creates 112 unequal length versions of the UCR datasets, compares methods of
preprocessing unequal length series for classification, and adapts the components of
HIVE-COTE 2.0 to handle unequal length series internally.

The code and results for the paper are in the
[unequal-length-hc2-2026](https://github.com/time-series-machine-learning/unequal-length-hc2-2026)
repository:

- [Results](https://github.com/time-series-machine-learning/unequal-length-hc2-2026/tree/main/results)
  for each preprocessing method and for the classifiers handling unequal length
  internally
- [Datasets](https://github.com/time-series-machine-learning/unequal-length-hc2-2026#datasets)
  used in the experiments, which are also available from
  [Zenodo](https://doi.org/10.5281/zenodo.18300287)
- [Running the experiments](https://github.com/time-series-machine-learning/unequal-length-hc2-2026#running-the-experiments)
  with the versions used in the paper
- [Classifier and preprocessing configurations](https://github.com/time-series-machine-learning/unequal-length-hc2-2026/blob/main/set_unequal_classifier.py)
  and the [unequal length HC2 code](https://github.com/time-series-machine-learning/unequal-length-hc2-2026/tree/main/algorithm_code)
- [Example notebook](https://github.com/time-series-machine-learning/unequal-length-hc2-2026/blob/main/basics.ipynb)
  using the unequal length HC2
